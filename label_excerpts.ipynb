# Labeling 19 excerpts

```python
import os
import json
from tqdm import tqdm  # progress bar

path_to_excerpts_folder = "tiny-dataset-first-iteration"
labels_output_file = "labeled_excerpts.jsonl"

# 1. Collect all .musicxml files from the folder
musicxml_files = [
    os.path.join(path_to_excerpts_folder, f)
    for f in os.listdir(path_to_excerpts_folder)
    if f.lower().endswith(".musicxml")
]

# 2. Open output file once and iterate with progress bar
with open(labels_output_file, 'a', encoding='utf-8') as f:
    for file_path in tqdm(musicxml_files, desc="Labeling MusicXML files"):
        try:
            # Load and convert to TinyNotation
            score = converter.parse(file_path, forceSource=True).explode()
            tinyscore = score_to_tinyscore(score)

            # Extract metadata
            score_info = ScoreInfoExtractor(score)

            # First-pass labeling
            prompt = LABEL_GENERATION_PROMPT.build_prompt(
                KEY=score_info.get_key(),
                METER=score_info.get_time_signatures(),
                NUMBER=score_info.get_measures(),
                PASTE_TINYNOTATION_HERE=tinyscore,
                with_system_prompt=True
            )
            labeling_response = generate_response(prompt)

            # Validation pass
            validation_prompt = LABEL_VALIDATION_PROMPT.build_prompt(
                KEY=score_info.get_key(),
                METER=score_info.get_time_signatures(),
                PASTE_TINYNOTATION_HERE=tinyscore,
                FIRST_PASS_JSON=labeling_response
            )
            validation_response = generate_response(validation_prompt)

            # Save results
            metadata = {
                "piece_name": score_info.get_piece_name(),
                "composer": score_info.get_composer(),
                "time_signature": score_info.get_time_signatures(),
                "key": score_info.get_key(),
                "measures": score_info.get_measures(),
                "file_path": file_path
            }
            output = {
                "metadata": metadata,
                "labels": validation_response.get('entities', [])
            }

            f.write(json.dumps(output, ensure_ascii=False) + "\n")

        except Exception as e:
            print(f"Failed to process {file_path}: {e}")
```
